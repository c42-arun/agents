```python
import truststore

truststore.inject_into_ssl()

from scripts import LeaseSoftClient

client = LeaseSoftClient.from_env()

print("LeaseSoft client configured with native certificate trust")
```

## Get titles

Set the LeaseSoft tenant schema, then call the `GetTitles` endpoint.

```python
# Replace this with the tenant schema you want to query.
tenant_schema = "IAF044T"
tenant_schema
```

```python
from html import escape
from IPython.display import HTML, display

if tenant_schema == "YOUR_TENANT_SCHEMA":
    raise ValueError("Replace YOUR_TENANT_SCHEMA before calling the API")

titles = client.get_titles(tenant_schema)
rows = "".join(
    f"<tr><td>{escape(str(title.get('code') or ''))}</td>"
    f"<td>{escape(str(title.get('description') or ''))}</td></tr>"
    for title in titles
)
table = (
    "<table><thead><tr><th>Code</th><th>Description</th></tr></thead>"
    f"<tbody>{rows}</tbody></table>"
)

print(f"Found {len(titles)} title values")
display(HTML(table))
```
